```python
import os
import matplotlib.pyplot as plt

def plot_frame(traj_point, target, save_path):
    lx, ly, rx, ry = traj_point

    fig = plt.figure(figsize=(2,2), dpi=64)

    # plot arms
    plt.scatter(lx, ly, c='blue', s=50)
    plt.scatter(rx, ry, c='green', s=50)# edgecolors='black', linewidth=0.5

    # plot target
    plt.scatter(target[0], target[1], c='red', marker='^', s=80)
    # FIXED axes (VERY IMPORTANT)
    plt.xlim(-0.8, 0.8)
    plt.ylim(-0.1, 0.8)

    plt.axis('off')

    fig.savefig(save_path)
    plt.close(fig)
```

```python
import pickle
import numpy as np

# load trajectories
with open("splined_trajectories_3.txt", "rb") as f:
    trajectories = pickle.load(f)

save_root = "trajectory_images"
os.makedirs(save_root, exist_ok=True)

for i, traj in enumerate(trajectories):

    traj_folder = os.path.join(save_root, f"traj_{i}")
    os.makedirs(traj_folder, exist_ok=True)

    # get target (same logic as before)
    left = traj[:, :2]
    right = traj[:, 2:]

    if np.max(np.linalg.norm(left - left[0], axis=1)) > np.max(np.linalg.norm(right - right[0], axis=1)):
        target = left[-1]
    else:
        target = right[-1]

    # generate frames
    for t, point in enumerate(traj):
        save_path = os.path.join(traj_folder, f"frame_{t}.png")
        plot_frame(point, target, save_path)
```
